```python
# --- 1. 定義とパス設定 ---
PREDICT_DIR = '../data/processed_test/'
RAW_DATA_PATH = '../data//test.csv' 
```

```python
### データの読み込み ###
df_test_original= pd.read_csv('test.csv')
df_test_original
```

```python
df_test_original
```

```python
#全部欠損値の行削除
df_test_original=df_test_original.dropna(axis=1,how='all')
df_test_original.info()
```

```python
#欠損値の数確認
df_test_original.isnull().sum()
```

```python
df_test=df_test_original.copy()
# 面積を数値型
df_test['面積（㎡）'] = df_test['面積（㎡）'].replace('2000㎡以上', 2000)
df_test['面積（㎡）']= pd.to_numeric(df_test['面積（㎡）'], errors='coerce')
#距離の数値変換と欠損値や？の削除、
df_test['最寄駅：距離（分）']= pd.to_numeric(df_test['最寄駅：距離（分）'], errors='coerce')
```

```python
# 削除対象の列
columns_to_drop = ['種類']

# 存在を確認してから削除（エラーを防ぐため）
existing_columns = [col for col in columns_to_drop if col in df_test.columns]

if existing_columns:
    df_test = df_test.drop(columns=existing_columns)
    print(f"以下の列を削除しました: {existing_columns}")
    print(f"現在のデータフレームの列数: {len(df_test.columns)}")

print(df_test.head())
```

```python
# --- 4. 処理後のデータを保存 ---
processed_file = 'test_cleaned.parquet'
processed_path = os.path.join(PREDICT_DIR, processed_file)

os.makedirs(PREDICT_DIR, exist_ok=True) # processed_testフォルダ作成

df_test.to_parquet(processed_path, index=False)
print(f"\n✅ クレンジング後のデータを {processed_path} (processed_test) に保存しました。")
```
