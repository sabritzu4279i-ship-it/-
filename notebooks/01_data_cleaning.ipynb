```python
file_list=glob.glob('train/*.csv')
original_df=pd.DataFrame()

for file in file_list:
    df_temp=pd.read_csv(file)
    original_df=pd.concat([original_df,df_temp])
```

```python
original_df
```

```python
#全部欠損値の行削除
original_df=original_df.dropna(axis=1,how='all')
original_df.info()
```

```python
#欠損値の数確認
original_df.isnull().sum()
```

```python
#統合したものをエクセル保存
#df.to_excel('data_merged.xlsx',index=None)
```

```python
df=original_df.copy()
# 面積を数値型
df['面積（㎡）'] = df['面積（㎡）'].replace('2000㎡以上', 2000)
df['面積（㎡）']= pd.to_numeric(df['面積（㎡）'], errors='coerce')
#距離の数値変換と欠損値や？の削除、
df['最寄駅：距離（分）']= pd.to_numeric(df['最寄駅：距離（分）'], errors='coerce')
df.dropna(subset=['最寄駅：距離（分）'], inplace=True)
#建蔽率、容積率の欠損値削除、
# '建ぺい率（％）'または'容積率（％）'のどちらかに欠損値がある行を削除
df.dropna(subset=['建ぺい率（％）', '容積率（％）'], inplace=True)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. '都道府県'列の件数を集計（件数が多い順にソート）
prefecture_counts = df['都道府県名'].value_counts()

# 2. 総件数を計算
total_count = len(df) # または df_test['都道府県'].count()

# 3. 割合（パーセント）を計算
prefecture_percentage = (prefecture_counts / total_count) * 100

# 4. 件数と割合をまとめたDataFrameを作成
summary_df = pd.DataFrame({
    '件数': prefecture_counts,
    '割合 (%)': prefecture_percentage.round(2)  # 小数点第2位まで表示
})

# 結果の表示
print("--- 都道府県別 件数と割合 ---")
print(f"データ総件数: {total_count} 件")
print("-" * 30)
print(summary_df)

# 5. グラフ表示（割合を反映）
# 割合をプロットすることで、全体の中で各都道府県がどれくらいの大きさを持つか視覚的に分かります。
plt.figure(figsize=(12, 7))
ax = prefecture_percentage.plot(kind='bar', color='darkblue')

# グラフのタイトルとラベルを設定
plt.title('都道府県別 割合 (%)', fontsize=16)
plt.xlabel('都道府県', fontsize=12)
plt.ylabel('割合 (%)', fontsize=12)

# X軸のラベル（都道府県名）を回転
plt.xticks(rotation=45, ha='right')

# 各棒の上に割合の値をテキストで表示
for i in ax.patches:
    # 棒の中心のX座標と、棒の高さ（割合）を取得
    x = i.get_x() + i.get_width() / 2
    y = i.get_height()
    # テキストを表示
    ax.annotate(f'{y:.2f}%', (x, y), ha='center', va='bottom',
                fontsize=9, color='black')

# グラフを表示
plt.tight_layout()
plt.show()
```

```python
# 削除対象の列
columns_to_drop = ['ID', '種類']

# 存在を確認してから削除（エラーを防ぐため）
existing_columns = [col for col in columns_to_drop if col in df.columns]

if existing_columns:
    df = df.drop(columns=existing_columns)
    print(f"以下の列を削除しました: {existing_columns}")
    print(f"現在のデータフレームの列数: {len(df.columns)}")

print(df.head())
```

```python
# --- 3. 処理後のデータを保存 ---
# 保存先のフォルダとファイル名を指定
processed_dir = '../data/processed/'
processed_file = 'cleaned_data.parquet'
processed_path = os.path.join(processed_dir, processed_file)

# processedフォルダがなければ作成する
os.makedirs(processed_dir, exist_ok=True)

# Parquet形式で保存（次の工程に引き継ぐ）
df.to_parquet(processed_path, index=False)

print(f"\nクレンジング後のデータを {processed_path} に保存しました。")
```
